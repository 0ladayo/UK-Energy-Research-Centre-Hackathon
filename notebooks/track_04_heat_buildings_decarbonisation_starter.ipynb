### 2. Loading Track 04 Datasets
We load:
1. **Electrification of Heat (EoH) Performance Summary**: Field-trial measured heat pump efficiencies (SPF H4 / COP), heating output, and electrical consumption.
2. **IDEAL Household IoT Sensor Telemetry**: High-resolution indoor temperatures and relative humidity across rooms.
3. **Open EPC Domestic England Sample**: Energy efficiency ratings (current vs. potential), floor area, wall/roof insulation.
4. **UK Cold Snap Winter Weather 2022**: Hourly outdoor temperatures and humidity during the severe December 2022 cold spell.


```python
data_dir = "Heat, Buildings & Decarbonisation"

# 1. Electrification of Heat (EoH) Heat Pump Performance
eoh_path = os.path.join(data_dir, "eoh_interim_performance_summary.csv")
df_eoh = pd.read_csv(eoh_path)
print("EoH Heat Pump Trial Shape:", df_eoh.shape)
display(df_eoh.head(3))
```

```python
# 2. IDEAL Household Indoor Temperature & Humidity
ideal_path = os.path.join(data_dir, "ideal_household_temp_humidity_sample.csv")
df_ideal = pd.read_csv(ideal_path)
df_ideal["time"] = pd.to_datetime(df_ideal["time"])
print("IDEAL IoT Telemetry Shape:", df_ideal.shape)
display(df_ideal.head(3))
```

```python
# 3. Open EPC Sample England
epc_path = os.path.join(data_dir, "open_epc_domestic_sample_england.csv")
df_epc = pd.read_csv(epc_path)
print("EPC Sample Shape:", df_epc.shape)
display(df_epc.head(3))
```

```python
# 4. UK Winter Cold Snap Hourly Weather (Dec 2022)
weather_path = os.path.join(data_dir, "uk_cold_snap_winter_weather_hourly.csv")
df_weather = pd.read_csv(weather_path)
df_weather["time"] = pd.to_datetime(df_weather["time"])
print("Cold Snap Weather Shape:", df_weather.shape)
display(df_weather.head(3))
```

### 3. Exploratory Analysis & Heat Decarbonisation Insights
#### Analysis 1: Real-World Heat Pump Efficiency (SPF H4) by Technology Type


```python
# Check available heat pump types and performance metrics
print("EoH Columns:", df_eoh.columns.tolist()[:8])

spf_col = [c for c in df_eoh.columns if "SPF" in c or "spf" in c.lower() or "cop" in c.lower()][0]
tech_col = [c for c in df_eoh.columns if "Type" in c or "type" in c.lower() or "Technology" in c][0]

plt.figure(figsize=(10, 5))
df_eoh.boxplot(column=spf_col, by=tech_col, grid=True, patch_artist=True)
plt.title("Heat Pump Seasonal Performance Factor (SPF) by Technology Type")
plt.suptitle("")
plt.xlabel("Heat Pump Technology")
plt.ylabel("Measured SPF (Efficiency Ratio)")
plt.xticks(rotation=15)
plt.show()
```

#### Analysis 2: Indoor Comfort vs. Sub-Zero Outdoor Cold Snap (December 2022)


```python
# Plot outdoor temperature profile alongside indoor sensor telemetry
plt.figure(figsize=(12, 5))
plt.plot(df_weather["time"], df_weather["outdoor_temp_C"], label="Outdoor Temperature (°C)", color="blue", linewidth=1.5)
plt.axhline(0, color="red", linestyle="--", alpha=0.7, label="Freezing Threshold (0°C)")
plt.title("December 2022 UK Cold Snap: Hourly Outdoor Temperature Profile")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

#### Analysis 3: EPC Current vs. Potential Energy Efficiency Rating


```python
# Compare current vs potential rating distribution
if "CURRENT_ENERGY_RATING" in df_epc.columns and "POTENTIAL_ENERGY_RATING" in df_epc.columns:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    df_epc["CURRENT_ENERGY_RATING"].value_counts().sort_index().plot(
        kind="bar", ax=axes[0], color="#f0ad4e", edgecolor="black"
    )
    axes[0].set_title("Current EPC Energy Rating")
    axes[0].set_xlabel("Rating Band")
    axes[0].set_ylabel("Property Count")
    
    df_epc["POTENTIAL_ENERGY_RATING"].value_counts().sort_index().plot(
        kind="bar", ax=axes[1], color="#5cb85c", edgecolor="black"
    )
    axes[1].set_title("Potential EPC Energy Rating After Retrofit")
    axes[1].set_xlabel("Rating Band")
    
    plt.tight_layout()
    plt.show()
```

### 4. Hackathon Starter Ideas & Challenge Questions
- **Thermal Decay & Envelope Efficiency**: Measure the rate at which indoor temperature drops when heating is off during freezing weather to estimate building thermal loss coefficients.
- **COP Sizing & Peak Load Modeling**: Model how heat pump coefficient of performance (COP) degrades under sub-zero ambient conditions and what peak electrical grid demand this implies.
- **Retrofit ROI & Carbon Abatement**: Estimate the carbon and cost savings of moving properties from EPC Band D/E to Band B/C using heat pumps combined with fabric insulation.
